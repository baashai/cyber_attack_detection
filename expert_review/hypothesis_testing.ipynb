```python
# Load the dataset
import pandas as pd
import scipy.stats as stats

file_path = "/content/modified_dataset (1).csv"
df = pd.read_csv(file_path)

# Display dataset information and first few rows
df.info(), df.head()
```

```python
# 1. T-Test: AI's impact on Attack Detection and Security Breach Prevention
t_stat_attacks, p_value_attacks = stats.ttest_ind(df["Attacks Detected by AI"], df["Security Breaches Prevented"])

# 2. ANOVA: AI's Effectiveness Across Different Industry Sectors
anova_stat_sec, p_value_anova_sec = stats.f_oneway(*[
    df[df["Industry Sector"] == sector]["Security Breaches Prevented"]
    for sector in df["Industry Sector"].unique()
])

# 3. Chi-Square Test: AI Adoption vs. Data Security Improvements
contingency_table_security = pd.crosstab(df["AI Adoption Level"], df["Post-AI Attack Reduction (%)"] > 50)
chi2_stat_sec, p_value_chi2_sec, dof_sec, expected_sec = stats.chi2_contingency(contingency_table_security)

# Output results
{
    "T-Test (AI vs. Security Prevention)": {"t-statistic": t_stat_attacks, "p-value": p_value_attacks},
    "ANOVA (Industry-wise Security Improvement)": {"F-statistic": anova_stat_sec, "p-value": p_value_anova_sec},
    "Chi-Square (AI Adoption vs. Security Improvement)": {"chi2-statistic": chi2_stat_sec, "p-value": p_value_chi2_sec}
}
```

```python
# Decision on hypothesis
alpha = 0.05  # Significance level
if p_value_attacks < alpha:
    print("Fail to Reject H₀: AI significantly improves cybersecurity.")
else:
    print("Reject H₀: No significant improvement detected.")
```
